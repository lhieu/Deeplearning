```python
import os
import shutil
import zipfile
from google.colab import files, drive

# 1. Đảm bảo Drive đã được mount
if not os.path.exists('/content/drive/MyDrive'):
    drive.mount('/content/drive')

output_zip = "/content/Fruits_classification_trained.zip"
drive_dest = "/content/drive/MyDrive/Fruits_classification_trained.zip"

files_to_pack = [
    "backbone.py",
    "data_loader.py",
    "train.py",
    "best_custom_fruitnet.pth"
]

# 2. Đóng gói các file
print("--> Bắt đầu nén file...")
with zipfile.ZipFile(output_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for file_name in files_to_pack:
        # Kiểm tra ở thư mục hiện tại hoặc thư mục cha /content/
        file_path = None
        if os.path.exists(file_name):
            file_path = file_name
        elif os.path.exists(os.path.join("/content", file_name)):
            file_path = os.path.join("/content", file_name)
        elif os.path.exists(os.path.join("/content/Fruits_classification", file_name)):
            file_path = os.path.join("/content/Fruits_classification", file_name)

        if file_path:
            zipf.write(file_path, arcname=file_name)
            print(f"  + Đã thêm: {file_name}")
        else:
            print(f"  ! Cảnh báo: Không tìm thấy {file_name}")

zip_size_mb = os.path.getsize(output_zip) / (1024 * 1024)
print(f"--> Tạo file zip thành công ({zip_size_mb:.2f} MB)")

# 3. Lưu vào Google Drive
shutil.copy(output_zip, drive_dest)
print(f"--> [OK] Đã lưu vào Drive: {drive_dest}")

# 4. Kích hoạt tải về máy tính qua trình duyệt
print("--> Đang khởi tạo tải về máy tính...")
files.download(output_zip)
```
